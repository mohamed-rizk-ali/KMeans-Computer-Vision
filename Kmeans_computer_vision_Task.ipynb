```python

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

X, _ = make_blobs(
    n_samples=300,
    centers=4,
    n_features=2,
    cluster_std=1.3,
    center_box=(-12, 12),
    random_state=24
)

plt.figure(figsize=(8, 5))

plt.scatter(X[:, 0], X[:, 1], s=40)

plt.title("Generated Virtual Data")
plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")
plt.grid(alpha=0.3)
plt.show()

kmeans = KMeans(
    n_clusters=4,
    init='k-means++',
    n_init=20,
    max_iter=300,
    random_state=24
)

kmeans.fit(X)

labels = kmeans.labels_
centroids = kmeans.cluster_centers_

plt.figure(figsize=(10, 6))

plt.scatter(
    X[:, 0],
    X[:, 1],
    c=labels,
    cmap='plasma',
    s=50
)

plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    marker='X',
    c='black',
    s=250,
    label='Centroids'
)

plt.title("K-Means Clustering using Scikit-learn")
plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

print("Total Data Points:", len(X))
print("Number of Clusters:", kmeans.n_clusters)
print("\nCluster Centers:\n", centroids)
print("\nInertia:", kmeans.inertia_)
print("Number of Iterations:", kmeans.n_iter_)

for i in range(kmeans.n_clusters):
    count = np.sum(labels == i)
    print(f"Cluster {i + 1}: {count} points")
```
